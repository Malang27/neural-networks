# Exercie 1

```python
import math

class Neuron:
    def __init__(self, weight1, weight2, bias):
        self.weight1 = weight1
        self.weight2 = weight2
        self.bias = bias

    def sigmoid(self, x):
        return 1 / (1 + math.exp(-x))

    def feedforward(self, x1, x2):
        # Étape 1 : Multiplier les entrées par les poids
        weighted_sum = (x1 * self.weight1) + (x2 * self.weight2) + self.bias
        
        # Étape 2 : Appliquer la fonction d'activation sigmoïde
        y = self.sigmoid(weighted_sum)
        
        return y

# Exemple d'utilisation
neuron = Neuron(0, 1, 4)
output = neuron.feedforward(2, 3)
print(output)  # Cela devrait imprimer une valeur proche de 0.999
```

# Exercice 3

```python
import numpy as np
from sklearn.metrics import log_loss

def log_loss_custom(y_true, y_pred, epsilon=1e-15):
    
    # Clip predictions to avoid log(0) or log(1)
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    
    # Calculate the log loss for each sample
    losses = y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred)
    
    # Return the negative mean
    return -np.mean(losses)

# Test data
y_true = np.array([0, 1, 1, 0, 1])
y_pred = np.array([0.1, 0.8, 0.6, 0.5, 0.3])

# Calculate using our custom function
custom_loss = log_loss_custom(y_true, y_pred)

# Calculate using sklearn's function
sklearn_loss = log_loss(y_true, y_pred)

print(f"Custom Log Loss: {custom_loss:.6f}")
print(f"Sklearn Log Loss: {sklearn_loss:.6f}")
```

# Exercice 4

```python
import numpy as np

class Neuron:
    def __init__(self, weight1, weight2, bias):
        self.weight1 = weight1
        self.weight2 = weight2
        self.bias = bias

    def sigmoid(self, x):
        return 1 / (1 + math.exp(-x))

    def feedforward(self, x1, x2):
        # Étape 1 : Multiplier les entrées par les poids
        weighted_sum = (x1 * self.weight1) + (x2 * self.weight2) + self.bias
        
        # Étape 2 : Appliquer la fonction d'activation sigmoïde
        y = self.sigmoid(weighted_sum)
        
        return y
 
class OurNeuralNetwork:
    def __init__(self, neuron_h1, neuron_h2, neuron_o1):
        self.h1 = neuron_h1
        self.h2 = neuron_h2
        self.o1 = neuron_o1
    
    def feedforward(self, x1, x2):
        h1_output = self.h1.feedforward(x1, x2)
        h2_output = self.h2.feedforward(x1, x2)
        return self.o1.feedforward(h1_output, h2_output)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def log_loss_custom(y_true, y_pred, epsilon=1e-15):
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
    losses = y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred)
    return -np.mean(losses)

# Create the dataset
data = [
    {"name": "Bob", "math": 12, "chemistry": 15, "exam_success": 1},
    {"name": "Eli", "math": 10, "chemistry": 9, "exam_success": 0},
    {"name": "Tom", "math": 18, "chemistry": 18, "exam_success": 1},
    {"name": "Ryan", "math": 13, "chemistry": 14, "exam_success": 1}
]

# Create the neural network
neuron_h1 = Neuron(0.05, 0.001, 0)
neuron_h2 = Neuron(0.02, 0.003, 0)
neuron_o1 = Neuron(2, 0, 0)
network = OurNeuralNetwork(neuron_h1, neuron_h2, neuron_o1)

# Step 1: Compute outputs for each student
outputs = []
y_true = []

print("Neural Network Outputs:")
print("----------------------")
for student in data:
    # Get network output
    output = network.feedforward(student["math"], student["chemistry"])
    
    outputs.append(output)
    y_true.append(student["exam_success"])
    print(f"{student['name']}: {output}")

# Convert to numpy arrays
y_true = np.array(y_true)
y_pred = np.array(outputs)

# Step 2: Compute log loss
loss = log_loss_custom(y_true, y_pred)
print("\nLog Loss:")
print(f"Loss value: {loss}")
```

# Exercie 5

```python
class Neuron:
    def __init__(self, weight1, weight2, bias, regression):
        self.weights_1 = weight1
        self.weights_2 = weight2
        self.bias = bias
        self.regression = regression
    def sigmoid(self, x):
       return 1 / (1 + math.exp(-x))

    def feedforward(self, x1, x2):
        
        # Calculate weighted sum plus bias
        total = (self.weights_1 * x1 + self.weights_2 * x2 + self.bias)
        # If regression, use identity function (return as is)
        # Otherwise, use ReLU activation
        
        if self.regression:
            return total
        else:
            return self.sigmoid(total)  # ReLU activation

class OurNeuralNetwork:
    def __init__(self, neuron_h1, neuron_h2, neuron_o1):
        self.h1 = neuron_h1
        self.h2 = neuron_h2
        self.o1 = neuron_o1
    
    def feedforward(self, x1, x2):
        h1_output = self.h1.feedforward(x1, x2)
        h2_output = self.h2.feedforward(x1, x2)
        return self.o1.feedforward(h1_output, h2_output)
    
# Question 1: Testing single neuron
neuron = Neuron(0, 1, 4, True)
output = neuron.feedforward(2, 3)
print(f"Question 1 output: {output}")  # Should be 7

# Question 2: Testing neural network for physics grade prediction
data = [
    {"name": "Bob", "math": 12, "chemistry": 15, "physics": 16},
    {"name": "Eli", "math": 10, "chemistry": 9, "physics": 10},
    {"name": "Tom", "math": 18, "chemistry": 18, "physics": 19},
    {"name": "Ryan", "math": 13, "chemistry": 14, "physics": 16}
]

# Create neural network (hidden neurons use ReLU, output neuron uses identity)
neuron_h1 = Neuron(0.05, 0.001, 0, False)  # Hidden neurons use ReLU
neuron_h2 = Neuron(0.002, 0.003, 0, False)  # Hidden neurons use ReLU
neuron_o1 = Neuron(2, 7, 10, True)  # Output neuron uses identity function
network = OurNeuralNetwork(neuron_h1, neuron_h2, neuron_o1)

# Calculate predictions and store true values
predictions = []
true_values = []

print("\nQuestion 2 outputs:")
for student in data:
    prediction = network.feedforward(student["math"], student["chemistry"])
    predictions.append(prediction)
    true_values.append(student["physics"])
    print(f"{student['name']}: {prediction}")

# Question 3: Calculate MSE
predictions = np.array(predictions)
true_values = np.array(true_values)
mse = np.mean((predictions - true_values) ** 2)
print(f"\nQuestion 3 - MSE: {mse}")
```
